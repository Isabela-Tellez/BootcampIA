##Ejemplo sencillo de implementación práctica
El siguiente fragmento de código ilustra cómo instanciar y aplicar las estrategias de imputación sobre una matriz de datos genérica utilizando Scikit-Learn:



```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer

# Creación de un DataFrame con valores faltantes
data = {
    'Edad': [23, 45, np.nan, 35, 60],
    'Ingresos': [21000, np.nan, 45000, 38000, 90000]
}
df = pd.DataFrame(data)

# Instanciación de imputadores
imputador_simple = SimpleImputer(strategy='median')
imputador_knn = KNNImputer(n_neighbors=2, weights='uniform')

# Ejemplo de ajuste y transformación
datos_imputados_provisionales = imputador_knn.fit_transform(df)
```
